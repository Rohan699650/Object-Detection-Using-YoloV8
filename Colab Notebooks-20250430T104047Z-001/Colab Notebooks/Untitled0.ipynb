```python
from ultralytics import YOLO

# Load the YOLOv8 Nano model
model = YOLO('yolov8n.yaml')

# Train the model for 10 epochs and save results locally in '/content/yolov8_local'
model.train(
    data='/content/drive/MyDrive/UAV/FOGGY (1)/data.yaml',  # Path to your data.yaml
    epochs=50,                  # Number of epochs
    project='/content/yolov8_local',  # Save results locally in this folder
    name='run1'                 # Subfolder for this specific training run
)

# Check the weights folder to verify
!ls /content/yolov8_local/run1/weights
```
