```python
import os
import cv2
from ultralytics import YOLO

# Define the video path (replace with your video file path)
video_path = '/content/test_videos/mist2.mp4'  # Update with the specific video path

# Define the output directory
output_dir = '/content/detected_videos'
os.makedirs(output_dir, exist_ok=True)

# Load the trained YOLO model
model = YOLO('/content/drive/MyDrive/updated_weights_epoch_35.pt')  # Replace with your model path

# Open the video using OpenCV
cap = cv2.VideoCapture(video_path)
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))

# Define output video writer
output_video_path = os.path.join(output_dir, os.path.basename(video_path))
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

print(f"Processing video: {video_path}")

frame_count = 0
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break  # End of video

    # Perform object detection on the current frame
    results = model.predict(source=frame, stream=True)  # Use stream=True for frame processing
    for result in results:
        annotated_frame = result.plot()  # Annotated frame with predictions
        out.write(annotated_frame)

    frame_count += 1
    if frame_count % 100 == 0:
        print(f"Processed {frame_count} frames...")

# Release resources
cap.release()
out.release()
print(f"Processed video saved to: {output_video_path}")
```

```python
from collections import Counter

# Initialize the Counter for detected classes
detected_classes = Counter()

# Inside the detection loop
for result in results:
    for box in result.boxes:
        class_id = int(box.cls)  # Class ID
        class_name = model.names[class_id]  # Class name
        detected_classes[class_name] += 1  # Increment count
```
