```python
import cv2
import numpy as np
import os
from google.colab.patches import cv2_imshow

def apply_fog(image, intensity=0.6):
    """
    Apply fog effect to an image.
    :param image: Input image
    :param intensity: Intensity of the fog effect (0 to 1, higher = denser fog)
    :return: Foggy image
    """
    h, w, _ = image.shape
    fog = np.random.normal(255 * intensity, 35, (h, w, 3)).astype(np.uint8)  # Increased fog variability
    foggy_image = cv2.addWeighted(image, 1 - intensity, fog, intensity, 0)
    return foggy_image

def process_dataset(input_folder, output_folder, intensity=0.6):
    """
    Apply fog effect to all images in a dataset folder and save them to a new folder.
    :param input_folder: Path to the folder containing original images
    :param output_folder: Path to the folder where foggy images will be saved
    :param intensity: Intensity of the fog effect
    """
    # Create the output folder if it doesn't exist
    os.makedirs(output_folder, exist_ok=True)

    # Process each image in the input folder
    for filename in os.listdir(input_folder):
        if filename.lower().endswith(('.png', '.jpg', '.jpeg')):  # Check for valid image files
            image_path = os.path.join(input_folder, filename)
            image = cv2.imread(image_path)

            if image is not None:
                foggy_image = apply_fog(image, intensity)
                output_path = os.path.join(output_folder, filename)
                cv2.imwrite(output_path, foggy_image)
                print(f"Processed: {filename}")
            else:
                print(f"Error reading file: {filename}")
        else:
            print(f"Skipping non-image file: {filename}")

# Paths for dataset
train_input_folder = "/content/data/images/train"  # Path to training images
validation_input_folder = "/content/data/images/val"  # Path to validation images

train_output_folder = "/content/drive/MyDrive/UAV/FOGGY/foggy_train"  # Path to save foggy training images
validation_output_folder = "/content/drive/MyDrive/UAV/FOGGY/foggy_validation"  # Path to save foggy validation images

# Fog intensity
fog_intensity = 0.6  # Adjust fog intensity as needed

# Process train and validation datasets
print("Processing training images...")
process_dataset(train_input_folder, train_output_folder, fog_intensity)

print("Processing validation images...")
process_dataset(validation_input_folder, validation_output_folder, fog_intensity)
```
